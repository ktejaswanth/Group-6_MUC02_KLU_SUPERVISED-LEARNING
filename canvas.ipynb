# CBIM Problem Canvas — Credit Pre-Screening Engine

## 1. Situation (S)

A mid-size private bank provides unsecured personal loans to customers across different employment types and city tiers. It receives approximately 3,000 applications each month, and every application currently requires around 25 minutes of manual credit review.

## 2. Complication — Pain Point

The bank's manual credit-screening process consumes significant analyst time, even though close to half of applications are eventually declined. This creates an opportunity to reduce repetitive review work, but the bank cannot simply automate declines based on predictions: incorrectly rejecting applicants who would have been approved could result in lost customers, unfair outcomes, and reduced trust in the lending process. The challenge is to improve screening efficiency without increasing the risk of wrongly declining applicants.

## 3. Question (Q)

How can the bank use historical loan-application data to identify high-confidence potential declines, reduce manual review workload, and keep historically approved applicants to no more than 5% of all auto-declined applications?

## 4. SSOT — Single Source of Truth

The provided `MUC02_Credit_PreScreening_Dataset.csv`, containing historical loan-application details and the credit team's recorded final decision in the `decision` column. This dataset will be the basis for analysing historical approval patterns and training and evaluating the pre-screening model.

## 5. Success Definition

Identify a probability threshold that satisfies the assignment's maximum 5% wrong-decline limit while auto-declining the largest number of applications among the tested thresholds that meet the limit. Quantify the potential monthly reduction in manual reviews and analyst hours, and examine whether the model's outcomes raise fairness concerns.

## 6. Primary Stakeholder

The Credit Policy Analyst, who needs evidence-based recommendations to present to the Head of Retail Lending. Credit analysts are affected by workload changes, while loan applicants are affected by automated screening decisions.
